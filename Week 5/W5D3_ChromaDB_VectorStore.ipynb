```python
import chromadb
from sentence_transformers import SentenceTransformer
```

```python
client = chromadb.Client()

print("Client Created")
```

```python
collection = client.create_collection("student_notes")

print("Collection Created")
```

```python
documents = [
    "Python is a programming language.",
    "Machine learning is a branch of artificial intelligence.",
    "Deep learning uses neural networks.",
    "ChromaDB is a vector database.",
    "Embeddings convert text into vectors.",
    "Git is used for version control.",
    "Docker creates containers.",
    "FastAPI builds APIs.",
    "NumPy is used for numerical computing.",
    "Pandas is used for data analysis.",
    "Matplotlib creates plots.",
    "Scikit-learn provides machine learning algorithms.",
    "Ollama runs LLMs locally.",
    "Large Language Models generate text.",
    "Semantic search uses embeddings.",
    "Metadata stores additional information.",
    "PDF files contain text and images.",
    "LangChain helps build AI applications.",
    "Retrieval Augmented Generation improves answers.",
    "Artificial Intelligence is transforming industries."
]
```

```python
ids = [str(i) for i in range(len(documents))]
```

```python
metadata = [
    {"topic":"python"},
    {"topic":"ai"},
    {"topic":"ai"},
    {"topic":"database"},
    {"topic":"database"},
    {"topic":"git"},
    {"topic":"docker"},
    {"topic":"api"},
    {"topic":"python"},
    {"topic":"python"},
    {"topic":"python"},
    {"topic":"ai"},
    {"topic":"llm"},
    {"topic":"llm"},
    {"topic":"database"},
    {"topic":"database"},
    {"topic":"pdf"},
    {"topic":"llm"},
    {"topic":"rag"},
    {"topic":"ai"}
]
```

```python
collection.add(
    ids=ids,
    documents=documents,
    metadatas=metadata
)

print("20 Documents Added")
```

```python
collection.count()
```

```python
results = collection.query(
    query_texts=["What is Artificial Intelligence?"],
    n_results=3
)

results["documents"]
```

```python
results = collection.query(
    query_texts=["Programming"],
    where={"topic":"python"},
    n_results=3
)

results["documents"]
```

```python
from pypdf import PdfReader

reader = PdfReader("sample.pdf")

text = ""

for page in reader.pages:
    if page.extract_text():
        text += page.extract_text()

print(text[:1000])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100
)

chunks = splitter.split_text(text)

len(chunks)
```

```python
pdf_collection = client.create_collection("pdf_data")
```

```python
pdf_collection.add(
    ids=[str(i) for i in range(len(chunks))],
    documents=chunks
)

print("PDF Stored")
```

```python
query = "What is the main topic?"

results = pdf_collection.query(
    query_texts=[query],
    n_results=3
)

top_chunks = results["documents"][0]

for i, chunk in enumerate(top_chunks, 1):
    print(f"Chunk {i}\n")
    print(chunk)
    print("-"*50)
```
