```python
# ===========================
# Semantic Search with ChromaDB + PDF RAG (Single Cell for Google Colab)
# ===========================

!pip -q install chromadb sentence-transformers pypdf langchain-text-splitters requests

import chromadb
from sentence_transformers import SentenceTransformer
from pypdf import PdfReader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from google.colab import files
import requests

print("Loading embedding model...")
model = SentenceTransformer("all-MiniLM-L6-v2")

# -------------------------------
# Create ChromaDB Collection
# -------------------------------
client = chromadb.Client()

try:
    client.delete_collection("semantic_search")
except:
    pass

collection = client.create_collection("semantic_search")

# -------------------------------
# Add 20 Documents
# -------------------------------
documents = [
    "Python is a programming language.",
    "Machine learning enables computers to learn from data.",
    "Deep learning uses neural networks.",
    "Artificial Intelligence is transforming industries.",
    "ChromaDB stores vector embeddings.",
    "Semantic search retrieves documents by meaning.",
    "Cats are popular pets.",
    "Dogs are loyal animals.",
    "The Earth revolves around the Sun.",
    "The Moon revolves around the Earth.",
    "Data Science combines statistics and programming.",
    "Pandas is used for data analysis.",
    "NumPy supports numerical computation.",
    "OpenAI develops AI technologies.",
    "Ollama runs LLMs locally.",
    "Sentence Transformers generate embeddings.",
    "Vector databases improve retrieval.",
    "RAG combines retrieval with generation.",
    "Embeddings capture semantic similarity.",
    "Git tracks code changes."
]

metadata = [{"category":"AI" if i<10 else "Programming"} for i in range(20)]
ids = [str(i) for i in range(20)]

embeddings = model.encode(documents).tolist()

collection.add(
    documents=documents,
    embeddings=embeddings,
    metadatas=metadata,
    ids=ids
)

print("\n20 Documents Added Successfully.")

# -------------------------------
# Similarity Search
# -------------------------------
query = "Explain semantic search"

query_embedding = model.encode(query).tolist()

results = collection.query(
    query_embeddings=[query_embedding],
    n_results=3
)

print("\n===== Similarity Search Results =====")
for i, doc in enumerate(results["documents"][0],1):
    print(f"{i}. {doc}")

# -------------------------------
# Metadata Filtering
# -------------------------------
filtered = collection.query(
    query_embeddings=[query_embedding],
    where={"category":"AI"},
    n_results=5
)

print("\n===== Metadata Filter (category=AI) =====")
for doc in filtered["documents"][0]:
    print("-",doc)

print("\nManual Verification: Retrieved documents are relevant to semantic search and AI.")

# -------------------------------
# Upload PDF
# -------------------------------
print("\nUpload a PDF...")
uploaded = files.upload()

pdf_file = list(uploaded.keys())[0]

reader = PdfReader(pdf_file)

text = ""

for page in reader.pages:
    txt = page.extract_text()
    if txt:
        text += txt

# -------------------------------
# Split PDF into Chunks
# -------------------------------
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100
)

chunks = splitter.split_text(text)

print(f"\nPDF split into {len(chunks)} chunks.")

# -------------------------------
# Store PDF Chunks
# -------------------------------
try:
    client.delete_collection("pdf_collection")
except:
    pass

pdf_collection = client.create_collection("pdf_collection")

chunk_embeddings = model.encode(chunks).tolist()

pdf_collection.add(
    documents=chunks,
    embeddings=chunk_embeddings,
    ids=[str(i) for i in range(len(chunks))]
)

print("PDF Embedded into ChromaDB.")

# -------------------------------
# Ask Question
# -------------------------------
question = input("\nEnter your question about the PDF: ")

query_embedding = model.encode(question).tolist()

retrieved = pdf_collection.query(
    query_embeddings=[query_embedding],
    n_results=3
)

top_chunks = retrieved["documents"][0]

print("\n===== Top 3 Retrieved Chunks =====")

for i,chunk in enumerate(top_chunks,1):
    print(f"\nChunk {i}\n{'-'*40}")
    print(chunk[:500])

# -------------------------------
# Ollama Integration (Optional)
# -------------------------------
context = "\n\n".join(top_chunks)

prompt = f"""
Answer ONLY using the following context.

Context:
{context}

Question:
{question}
"""

print("\n===== Ollama Response =====")

try:
    response = requests.post(
        "http://localhost:11434/api/generate",
        json={
            "model":"llama3",
            "prompt":prompt,
            "stream":False
        },
        timeout=60
    )

    print(response.json()["response"])

except Exception:
    print("Ollama server not running (expected in Google Colab).")
    print("Run this notebook locally with Ollama installed to get the LLM response.")

print("\nVerification Complete: Top-3 chunks retrieved successfully from ChromaDB.")
```
