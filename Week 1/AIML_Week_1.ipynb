Day 4 EDA

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create a sample dataset
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank"],
    "Age": [23, 25, 22, 24, 26, 23],
    "Marks": [85, 90, 78, 88, 95, 80],
    "Gender": ["F", "M", "M", "M", "F", "M"]
}

df = pd.DataFrame(data)
```

```python
print("\nDataset Info:")
print(df.info())

# Summary statistics
print("\nSummary Statistics:")
print(df.describe())

# Check for missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Correlation between numerical columns
print("\nCorrelation Matrix:")
print(df[["Age", "Marks"]].corr())
```

```python
plt.figure(figsize=(6,4))
plt.hist(df["Marks"], bins=5, color="skyblue", edgecolor="black")
plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()

# Bar Chart
plt.figure(figsize=(6,4))
plt.bar(df["Name"], df["Marks"], color="green")
plt.title("Marks of Students")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()

# Scatter Plot
plt.figure(figsize=(6,4))
plt.scatter(df["Age"], df["Marks"], color="red")
plt.title("Age vs Marks")
plt.xlabel("Age")
plt.ylabel("Marks")
plt.show()

# Box Plot
plt.figure(figsize=(6,4))
sns.boxplot(y=df["Marks"])
plt.title("Box Plot of Marks")
plt.show()

# Heatmap
plt.figure(figsize=(5,4))
sns.heatmap(df[["Age", "Marks"]].corr(),
            annot=True,
            cmap="Blues")
plt.title("Correlation Heatmap")
plt.show()
```

Day 5 EDA and Matplotlib and Seaborn

```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create sample dataset
data = {
    "Student": ["A", "B", "C", "D", "E", "F"],
    "Math": [85, 90, 78, 88, 95, 80],
    "Science": [90, 85, 82, 92, 88, 79],
    "Age": [20, 21, 19, 22, 20, 21]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)


# 1. Line Plot (Matplotlib)
plt.figure(figsize=(6,4))

plt.plot(df["Student"], df["Math"],
         marker="o",
         color="blue",
         label="Math Marks")

plt.title("Student Math Scores")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.legend()
plt.grid()
plt.show()


# 2. Bar Chart (Matplotlib)

plt.figure(figsize=(6,4))

plt.bar(df["Student"], df["Science"],
        color="green")

plt.title("Science Marks of Students")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()


# 3. Scatter Plot (Seaborn)
plt.figure(figsize=(6,4))

sns.scatterplot(
    x="Age",
    y="Math",
    data=df,
    hue="Student",
    s=100
)

plt.title("Age vs Math Marks")
plt.show()


# 4. Histogram (Seaborn)

plt.figure(figsize=(6,4))

sns.histplot(
    df["Math"],
    bins=5,
    color="purple"
)

plt.title("Distribution of Math Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()


# 5. Box Plot (Seaborn)
plt.figure(figsize=(6,4))

sns.boxplot(
    data=df[["Math", "Science"]]
)

plt.title("Marks Distribution")
plt.show()


# 6. Heatmap (Seaborn)
plt.figure(figsize=(5,4))

correlation = df[["Math", "Science", "Age"]].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()


# 7. Pair Plot (Seaborn)
sns.pairplot(df)

plt.show()
```
