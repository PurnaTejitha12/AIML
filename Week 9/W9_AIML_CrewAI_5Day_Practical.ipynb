```python
from crewai import Agent, Task, Crew, Process, LLM

llm = LLM(
    model="ollama/llama3.2:3b",
    base_url="http://localhost:11434"
)
```

```python
researcher = Agent(
    role="Researcher",
    goal="Research the given topic and identify important facts.",
    backstory="You are an expert research analyst.",
    llm=llm,
    verbose=True
)

writer = Agent(
    role="Writer",
    goal="Turn research findings into a clear article.",
    backstory="You are an experienced technical writer.",
    llm=llm,
    verbose=True
)

reviewer = Agent(
    role="Reviewer",
    goal="Check the article for accuracy, clarity and completeness.",
    backstory="You are a meticulous editor and fact checker.",
    llm=llm,
    verbose=True
)
```
