```python
import requests

def wikipedia_search(query):
    url = "https://en.wikipedia.org/w/api.php"

    # Add a User-Agent header to the request
    headers = {
        "User-Agent": "ColabNotebook/1.0 (https://colab.research.google.com; user@example.com)"
    }

    params = {
        "action": "query",
        "list": "search",
        "srsearch": query,
        "format": "json",
        "srlimit": 5
    }

    try:
        response = requests.get(url, headers=headers, params=params)
        response.raise_for_status()  # Raise an exception for HTTP errors

        data = response.json()

        results = []

        for item in data["query"]["search"]:
            results.append({
                "title": item["title"],
                "snippet": item["snippet"].replace("<span class=\"searchmatch\">", "")
                                             .replace("</span>", "")
            })

        return results
    except requests.exceptions.RequestException as e:
        print(f"Wikipedia API request failed: {e}")
        if hasattr(e, 'response') and e.response is not None:
            print(f"Status Code: {e.response.status_code}")
            print(f"Response Text: {e.response.text}")
        return []
    except Exception as e:
        print(f"An unexpected error occurred in wikipedia_search: {e}")
        return []
```

# Day 3 — Building a Research Crew — Multi-Agent Pipeline

```python
import nest_asyncio
nest_asyncio.apply()

from crewai import Agent, Task, Crew, Process, LLM

llm = LLM(
    model="ollama/llama3.2:3b",
    base_url="http://localhost:11434"
)
```

```python
researcher = Agent(
    role="Researcher",
    goal="Research the topic and provide useful information.",
    backstory="You are a research expert who finds important facts.",
    llm=llm,
    verbose=True
)

writer = Agent(
    role="Writer",
    goal="Turn research into a simple and clear article.",
    backstory="You are a professional technical writer.",
    llm=llm,
    verbose=True
)

reviewer = Agent(
    role="Reviewer",
    goal="Check the article for quality, clarity, and completeness.",
    backstory="You are an experienced editor and reviewer.",
    llm=llm,
    verbose=True
)
```

```python
research_task = Task(
    description=f"""
    Research the topic:
    "Benefits and challenges of Generative AI in software development."

    Use the following web search results:

    {web_data}

    Identify:
    - Major benefits
    - Common applications
    - Main challenges
    - Future impact
    Clearly separate information from unsupported assumptions.
    """,
    expected_output="A clear research summary with important points.",
    agent=researcher
)
```

```python
writing_task = Task(
    description="""
    Use the Researcher's findings to write a simple article
    about Generative AI in software development.

    Include headings and explain the benefits and challenges clearly.
    """,
    expected_output="A clear article of approximately 400-500 words.",
    agent=writer,
    context=[research_task]
)
```

```python
review_task = Task(
    description="""
    Review the Writer's article.

    Check:
    - Accuracy
    - Clarity
    - Completeness
    - Logical flow

    Give a score out of 10 and suggest improvements.
    """,
    expected_output="A review with strengths, weaknesses, score, and suggestions.",
    agent=reviewer,
    context=[writing_task]
)
```

```python
research_crew = Crew(
    agents=[researcher, writer, reviewer],
    tasks=[research_task, writing_task, review_task],
    process=Process.sequential,
    verbose=True
)
```
